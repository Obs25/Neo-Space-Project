```python
import requests
from datetime import datetime

asteroids_data = []
target = 10000
url = f"https://api.nasa.gov/neo/rest/v1/feed?start_date=2025-01-01&end_date=2025-01-08&api_key={API_KEY}"

while len(asteroids_data) < target:
    response = requests.get(url)
    data = response.json()
    details = data['near_earth_objects']
    
    for date, info in details.items():
        for i in info:
            diameter_info = i['estimated_diameter']['kilometers']
            close_approach = i['close_approach_data'][0] if i['close_approach_data'] else {}

            asteroids_data.append(dict(
                id=int(i['id']),
                name=i['name'],
                absolute_magnitude_h=float(i['absolute_magnitude_h']),
                estimated_diameter_min_km=float(diameter_info['estimated_diameter_min']),
                estimated_diameter_max_km=float(diameter_info['estimated_diameter_max']),
                is_potentially_hazardous_asteroid=i['is_potentially_hazardous_asteroid'],
                close_approach_date = datetime.strptime(close_approach['close_approach_date'], "%Y-%m-%d").date(),
                relative_velocity_kmph=float(close_approach.get('relative_velocity', {}).get('kilometers_per_hour')),
                astronomical=float(close_approach.get('miss_distance', {}).get('astronomical')),
                miss_distance_km=float(close_approach.get('miss_distance', {}).get('kilometers')),
                miss_distance_lunar=float(close_approach.get('miss_distance', {}).get('lunar')),
                orbiting_body=close_approach.get('orbiting_body')
            ))

            if len(asteroids_data) >= target:
                break
        if len(asteroids_data) >= target:
            break
    
    url = data['links']['next']
```

## 1. Count how many times each asteroid has approached Earth



```python
query1 = """
SELECT neo_reference_id, COUNT(*) AS approach_count
FROM astero.close_approach
GROUP BY neo_reference_id
ORDER BY approach_count DESC;
"""

curr.execute(query1)
result1 = curr.fetchall()
for row in result1:
    print(row)
```

## 2. Average velocity of each asteroid over multiple approaches



```python

query2 = """
SELECT neo_reference_id, AVG(relative_velocity_kmph) AS avg_velocity
FROM close_approach
GROUP BY neo_reference_id
ORDER BY avg_velocity DESC;
"""

curr.execute(query2)
result2 = curr.fetchall()
for row in result2:
    print(row)
```

## 3. List top 10 fastest asteroids

```python
# This gets the top 10 asteroids based on maximum recorded speed in any approach.
query3 = """
SELECT neo_reference_id, MAX(relative_velocity_kmph) AS max_velocity
FROM close_approach
GROUP BY neo_reference_id
ORDER BY max_velocity DESC
LIMIT 10;
"""

curr.execute(query3)
result3 = curr.fetchall()
for row in result3:
    print(row)
```

## 4. Find potentially hazardous asteroids that have approached Earth more than 3 times

```python

query4 = """
SELECT a.id, a.name, COUNT(c.neo_reference_id) AS approach_count
FROM asteroids a
JOIN close_approach c ON a.id = c.neo_reference_id
WHERE a.is_potentially_hazardous_asteroid = TRUE
GROUP BY a.id, a.name
HAVING approach_count > 3;
"""

curr.execute(query4)
result4 = curr.fetchall()
for row in result4:
    print(row)

```

## 5. Find the month with the most asteroid approaches

```python

query5 = """
SELECT MONTH(close_approach_date) AS approach_month, COUNT(*) AS total_approaches
FROM close_approach
GROUP BY approach_month
ORDER BY total_approaches DESC
LIMIT 1;
"""

curr.execute(query5)
result5 = curr.fetchall()
for row in result5:
    print(row)
```

## 6. Get the asteroid with the fastest ever approach speed

```python

query6 = """
SELECT neo_reference_id, close_approach_date, relative_velocity_kmph
FROM close_approach
ORDER BY relative_velocity_kmph DESC
LIMIT 1;
"""

curr.execute(query6)
result6 = curr.fetchall()
for row in result6:
    print(row)
```

## 7. Sort asteroids by maximum estimated diameter (descending)

```python
query7 = """
SELECT id, name, estimated_diameter_max_km
FROM asteroids
ORDER BY estimated_diameter_max_km DESC;
"""
curr.execute(query7)
result7 = curr.fetchall()
for row in result7:
    print(row)
```

## 8. Asteroids whose closest approach is getting nearer over time

```python
query8 = """
SELECT neo_reference_id, close_approach_date, miss_distance_km
FROM close_approach
ORDER BY neo_reference_id, close_approach_date;
"""
curr.execute(query8)
result8 = curr.fetchall()
for row in result8:
    print(row)
```

## 9. Display the name of each asteroid with date and miss distance of closest approach

```python
query9 = """
SELECT a.name, c.close_approach_date, c.miss_distance_km
FROM asteroids a
JOIN close_approach c ON a.id = c.neo_reference_id
ORDER BY c.miss_distance_km ASC;
"""
curr.execute(query9)
result9 = curr.fetchall()
for row in result9:
    print(row)
```

## 10. List names of asteroids that approached Earth with velocity > 50000 km/h

```python
query10 = """
SELECT DISTINCT a.name, c.relative_velocity_kmph
FROM asteroids a
JOIN close_approach c ON a.id = c.neo_reference_id
WHERE c.relative_velocity_kmph > 50000;
"""
curr.execute(query10)
result10 = curr.fetchall()
for row in result10:
    print(row)
```

## 11. Count how many approaches happened per month

```python
query11 = """
SELECT MONTH(close_approach_date) AS month, COUNT(*) AS total
FROM close_approach
GROUP BY month
ORDER BY total DESC;
"""
curr.execute(query11)
result11 = curr.fetchall()
for row in result11:
    print(row)
```

## 12. Find asteroid with the highest brightness (lowest magnitude value)

```python
query12 = """
SELECT id, name, absolute_magnitude
FROM asteroids
ORDER BY absolute_magnitude ASC
LIMIT 1;
"""
curr.execute(query12)
result12 = curr.fetchall()
for row in result12:
    print(row)
```

## 13. Get number of hazardous vs non-hazardous asteroids

```python
query13 = """
SELECT is_potentially_hazardous_asteroid, COUNT(*) AS count
FROM asteroids
GROUP BY is_potentially_hazardous_asteroid;
"""
curr.execute(query13)
result13 = curr.fetchall()
for row in result13:
    print(row)
```

## 14. Asteroids that passed closer than the Moon (<1 Lunar Distance)

```python
query14 = """
SELECT neo_reference_id, close_approach_date, miss_distance_lunar
FROM close_approach
WHERE miss_distance_lunar < 1;
"""
curr.execute(query14)
result14 = curr.fetchall()
for row in result14:
    print(row)
```
